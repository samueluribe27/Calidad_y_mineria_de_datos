# Diagnóstico de Calidad de Datos
## Dataset: Acuerdos Distritales - Cartagena de Indias

---

## 1. Completitud
El dataset cuenta con 435 registros y 10 columnas, sin ningún valor nulo en ninguna de las columnas. Todas las filas están completamente diligenciadas, lo que indica una completitud del 100%.

**Conclusión:** La completitud es ÓPTIMA.

---

## 2. Unicidad
No se encontraron registros duplicados en el dataset. La columna ID_acuerdo presenta 435 valores únicos, es decir, cada acuerdo tiene un identificador irrepetible.

**Conclusión:** La unicidad es ÓPTIMA.

---

## 3. Consistencia
Se identificaron inconsistencias en tres columnas:
- La columna **vigencia** está almacenada como texto con formato '2,008' en lugar de un número entero (2008).
- La columna **fec_sanc** está en formato string cuando debería ser de tipo datetime para permitir operaciones de fecha.
- La columna **estado_acuerdo** presenta inconsistencia en la capitalización: los valores 'Vigente' y 'Derogado' inician con mayúscula, mientras que 'modificado' está completamente en minúscula.

**Conclusión:** La consistencia es DEFICIENTE en 3 columnas y requiere corrección en la etapa de limpieza.

---

## 4. Exactitud
La columna **articulado** tiene un promedio de 12 artículos por acuerdo y una mediana de 5. Sin embargo, se detectaron 28 registros con más de 30 artículos, siendo el valor máximo 463 artículos (acuerdo AD-107-2022). Estos valores elevados no son necesariamente errores, ya que acuerdos de presupuesto o planes de desarrollo suelen contener muchos artículos.

**Conclusión:** La exactitud es ACEPTABLE. Los valores atípicos son justificables por la naturaleza del dato.

---

## 5. Validez
La columna **estado_acuerdo** contiene tres categorías: Vigente, Derogado y modificado. Esta última afecta a 11 registros (2.5% del dataset) y no está estandarizada respecto a las otras dos categorías. Los años de vigencia van de 2008 a 2025, todos dentro del rango esperado.

**Conclusión:** La validez es PARCIAL. Se recomienda unificar y estandarizar la categoría 'modificado'.

---

## 6. Oportunidad
El dataset fue actualizado en febrero de 2026 y contiene acuerdos desde 2008 hasta octubre de 2025. Los datos están vigentes y son pertinentes para el análisis actual.

**Conclusión:** La oportunidad es ÓPTIMA.

---

## Resumen del Diagnóstico

| Dimensión     | Estado      | Observación                              |
|---------------|-------------|------------------------------------------|
| Completitud   | Óptima   | 0 valores nulos                          |
| Unicidad      | Óptima   | 0 registros duplicados                   |
| Consistencia  | Deficiente | 3 columnas con formato incorrecto       |
| Exactitud     | Aceptable | Outliers justificables                  |
| Validez       | Parcial  | 'modificado' no estandarizado            |
| Oportunidad   | Óptima   | Datos actualizados hasta octubre de 2025 |

```python
df_limpio = df.copy()


df_limpio['vigencia'] = df_limpio['vigencia'].str.replace(',', '').astype(int)


df_limpio['fec_sanc'] = pd.to_datetime(df_limpio['fec_sanc'], dayfirst=True, errors='coerce')


df_limpio['estado_acuerdo'] = df_limpio['estado_acuerdo'].str.strip().str.capitalize()


df_limpio['tema'] = df_limpio['tema'].str.strip()


df_limpio = df_limpio[df_limpio['estado_acuerdo'].isin(['Vigente', 'Derogado'])]

print("Limpieza completada")
print(f"Registros después de limpieza: {df_limpio.shape[0]}")
print(df_limpio.dtypes)
print(df_limpio['estado_acuerdo'].value_counts())


df_limpio.to_csv('datos_limpios.csv', index=False)
print("\n Datos limpios guardados en: datos_limpios.csv")
```
